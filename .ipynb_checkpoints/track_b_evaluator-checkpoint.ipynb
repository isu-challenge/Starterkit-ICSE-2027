# Track B Evaluator

This notebook evaluates an in-car scene understanding system against the labeled images in the starter kit's root `data/` directory. It uses `Qwen/Qwen2.5-VL-3B-Instruct` to predict the scene features defined by the labels, then reports feature accuracy, scene exact-match accuracy, feature diversity, and latency.

Run the cells from top to bottom. Model weights are downloaded from Hugging Face the first time the model is loaded.

## 1. Configure the evaluation

Install the Track B dependencies before running this notebook:

```bash
pip install -r track_b/requirements.txt
```

[ISU-Test](https://github.com/ast-fortiss-tum/ISU-Test) can generate controllable synthetic in-car scenes for Track B development and training. Alternatively, participants can use the synthetic and real images already provided in the starter kit. In both cases, scenes should use the competition feature definitions and labels, and submitted systems should remain reproducible.

````python
import json
import sys
from pathlib import Path

import torch

NOTEBOOK_DIR = Path.cwd()
TRACK_B_DIR = NOTEBOOK_DIR if (NOTEBOOK_DIR / 'track_b_evaluator.py').exists() else NOTEBOOK_DIR / 'track_b'
if str(TRACK_B_DIR) not in sys.path:
    sys.path.insert(0, str(TRACK_B_DIR))

from track_b_evaluator import (
    DEFAULT_IGNORED_FEATURES,
    build_prompt,
    build_vocabulary,
    comparable_features,
    feature_metrics,
    load_dataset,
    normalize,
    parse_json_answer,
    query_model,
)

MODEL_NAME = 'Qwen/Qwen2.5-VL-3B-Instruct'
DATA_CANDIDATES = [
    NOTEBOOK_DIR / 'data',
    NOTEBOOK_DIR.parent / 'data',
    NOTEBOOK_DIR.parent.parent / 'data',
]
DATA_DIR = next((path for path in DATA_CANDIDATES if path.is_dir()), None)
if DATA_DIR is None:
    raise FileNotFoundError('Could not find the starter kit root data/ directory.')
IMAGE_KIND = 'both'
SCENE_LIMIT = None
OUTPUT_PATH = TRACK_B_DIR / 'track_b_evaluation_output' / 'report.json'
IGNORED_FEATURES = DEFAULT_IGNORED_FEATURES.copy()
DEVICE = 'cuda' if torch.cuda.is_available() else ('mps' if torch.backends.mps.is_available() else 'cpu')

print(f'Starter kit data directory: {DATA_DIR}')
print(f'Model: {MODEL_NAME}')
print(f'Device: {DEVICE}')
````

## 2. Load and inspect the labeled data

````python
scenes = load_dataset(DATA_DIR, IMAGE_KIND)
vocabulary = build_vocabulary(scenes, IGNORED_FEATURES)
PROMPT = build_prompt(vocabulary)

print(f'Loaded {len(scenes)} image-label pairs')
print('Features:', ', '.join(vocabulary))
print('Example prompt:')
print(PROMPT)
````

````python
inspection_rows = []
for scene_name, image_path, label in scenes:
    inspection_rows.append({
        'scene': scene_name,
        'image': str(image_path),
        **comparable_features(label, IGNORED_FEATURES),
    })

inspection_rows[:3]
````

## 3. Load Qwen2.5-VL and evaluate

Set `SCENE_LIMIT` to a small integer above for a smoke test. Leave it as `None` to evaluate every scene.

````python
from PIL import Image
from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration
import time

processor = AutoProcessor.from_pretrained(MODEL_NAME)
model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    MODEL_NAME,
    torch_dtype='auto',
    device_map=DEVICE,
).eval()
print('Model loaded')
````

````python
evaluation_scenes = scenes if SCENE_LIMIT is None else scenes[:SCENE_LIMIT]
prediction_rows = []

for index, (scene_name, image_path, label) in enumerate(evaluation_scenes, 1):
    image = Image.open(image_path).convert('RGB')
    started = time.perf_counter()
    raw_answer = query_model(model, processor, image, PROMPT, DEVICE)
    prediction = parse_json_answer(raw_answer) or {}
    expected = comparable_features(label, IGNORED_FEATURES)
    feature_matches = {
        key: normalize(prediction.get(key)) == normalize(value)
        for key, value in expected.items()
    }
    prediction_rows.append({
        'image': scene_name,
        'image_path': str(image_path),
        'expected': expected,
        'prediction': prediction,
        'feature_matches': feature_matches,
        'all_features_match': all(feature_matches.values()),
        'raw_answer': raw_answer,
        'inference_seconds': time.perf_counter() - started,
    })
    print(f'[{index}/{len(evaluation_scenes)}] {scene_name}', flush=True)
````

## 4. Calculate metrics and save the report

````python
metrics = {
    **feature_metrics(prediction_rows),
    'scene_exact_match': (
        sum(row['all_features_match'] for row in prediction_rows) / len(prediction_rows)
        if prediction_rows else None
    ),
    'latency_seconds_total': sum(row['inference_seconds'] for row in prediction_rows),
    'latency_seconds_mean': (
        sum(row['inference_seconds'] for row in prediction_rows) / len(prediction_rows)
        if prediction_rows else None
    ),
    'scenes': len(prediction_rows),
}

report = {
    'track': 'B',
    'model': MODEL_NAME,
    'device': DEVICE,
    'image_kind': IMAGE_KIND,
    'prompt': PROMPT,
    'features': vocabulary,
    'metrics': metrics,
    'predictions': prediction_rows,
}

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
OUTPUT_PATH.write_text(json.dumps(report, indent=2, ensure_ascii=True))
print(json.dumps(metrics, indent=2))
print(f'Saved report to {OUTPUT_PATH}')
````

````python
import pandas as pd

metrics_df = pd.Series(metrics, name='value').to_frame()
display(metrics_df)

print(f"Feature diversity: {metrics['feature_diversity']:.4f}")
print('Unique ground-truth values per feature:')
print(json.dumps(metrics['unique_values'], indent=2))

results_df = pd.DataFrame([
    {
        'scene': row['image'],
        'all_features_match': row['all_features_match'],
        'inference_seconds': row['inference_seconds'],
    }
    for row in prediction_rows
])
display(results_df.head())
````
